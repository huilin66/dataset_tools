```python
import os
from tqdm import tqdm
```

```python
input_label_dir = r'/localnvme/data/billboard/fused_data/data7961_mseg_c6_1023/labels'
output_label_dir = r'/localnvme/data/billboard/fused_data/data7961_mseg_c6_1029_abandonment_refine/labels'

single_risk_dir = r'/localnvme/data/billboard/fused_data/data7961_mseg_c6_1029_single_risk'
risk_list = ['d', 'b', 'a', 'c',]

os.makedirs(output_label_dir, exist_ok=True)
```

```python
def att_check(input_dir, output_dir):
    label_list = os.listdir(input_dir)
    for label_name in tqdm(label_list):
        input_path = os.path.join(input_dir, label_name)
        output_path = os.path.join(output_dir, label_name)
        with open(input_path, 'r') as f:
            lines = f.readlines()
            for idx, line in enumerate(lines):
                if line[6] == '2':
                    lines[idx] = lines[idx][0:6] + '0' + lines[idx][7:]
                else:
                    continue
        with open(output_path, 'w') as f:
            f.writelines(lines)
```

```python
att_check(input_label_dir, output_label_dir)
```

```python
def extract_single_risk(input_dir, output_dir, risk):
    save_dir = os.path.join(output_dir, risk, 'labels')
    os.makedirs(save_dir, exist_ok=True)
    label_list = os.listdir(input_dir)
    for label_name in tqdm(label_list, desc=risk):
        input_path = os.path.join(input_dir, label_name)
        output_path = os.path.join(save_dir, label_name)
        with open(input_path, 'r') as f:
            lines = f.readlines()
            new_lines = []
            for idx, line in enumerate(lines):
                parts = line.strip().split(' ')
                if risk == 'd':
                    parts[3], parts[4], parts[5] = '0', '0', '0'
                elif risk == 'b':
                    parts[2], parts[4], parts[5] = '0', '0', '0'
                elif risk == 'a':
                    parts[2], parts[3], parts[5] = '0', '0', '0'
                elif risk == 'c':
                    parts[2], parts[3], parts[4] = '0', '0', '0'
                new_line = ' '.join(parts) + '\n'
                new_lines.append(new_line)

        with open(output_path, 'w') as f:
            f.writelines(new_lines)
```

```python
for risk in risk_list:
    extract_single_risk(output_label_dir, single_risk_dir, risk)
```
