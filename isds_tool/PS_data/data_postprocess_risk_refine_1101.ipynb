```python
import os
import shutil
from tqdm import tqdm
```

```python
input_dir = r'/localnvme/data/billboard/fused_data/data7961_mseg_c5_l2_1031_v4'
input_label_dir = os.path.join(input_dir, 'labels')
input_image_dir = os.path.join(input_dir, 'images')
single_risk_dir = r'/localnvme/data/billboard/fused_data/data7961_mseg_c5_l2_1031_v4_single_risk'
risk_list = ['d', 'b', 'a', 'c',]
```

```python
def extract_single_risk(input_dir, output_dir, risk):
    save_dir = os.path.join(output_dir, risk, 'labels')
    os.makedirs(save_dir, exist_ok=True)
    label_list = os.listdir(input_dir)
    for label_name in tqdm(label_list, desc=risk):
        input_path = os.path.join(input_dir, label_name)
        output_path = os.path.join(save_dir, label_name)
        with open(input_path, 'r') as f:
            lines = f.readlines()
            new_lines = []
            for idx, line in enumerate(lines):
                parts = line.strip().split(' ')
                if risk == 'd':
                    parts[3], parts[4], parts[5] = '0', '0', '0'
                elif risk == 'b':
                    parts[2], parts[4], parts[5] = '0', '0', '0'
                elif risk == 'a':
                    parts[2], parts[3], parts[5] = '0', '0', '0'
                elif risk == 'c':
                    parts[2], parts[3], parts[4] = '0', '0', '0'
                new_line = ' '.join(parts) + '\n'
                new_lines.append(new_line)

        with open(output_path, 'w') as f:
            f.writelines(new_lines)
```

```python
for risk in risk_list:
    extract_single_risk(input_label_dir, single_risk_dir, risk)
```

```python
for risk in risk_list:
    output_image_dir = os.path.join(single_risk_dir, risk, 'images')
    print(f'copy {output_image_dir}')
    shutil.copytree(input_image_dir, output_image_dir)
```
