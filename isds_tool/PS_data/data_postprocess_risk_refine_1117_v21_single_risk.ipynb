```python
import os
import shutil
from att_tools import extract_single_risk_keep_single
from yolo_tools import copy_split,ref_split
```

```python
data_c5l2_dir =  r'/localnvme/data/billboard/all_data/mseg_c5_l2/data7961_mseg_c5_l2_1117_v21'
data_c5l2_dir_image = os.path.join(data_c5l2_dir, 'images')
data_c5l2_dir_label = os.path.join(data_c5l2_dir, 'labels')
single_risk_dir = data_c5l2_dir + '_single_risk'
risk_list = ['d', 'b', 'a', 'c']

class_c5_path = r'/localnvme/data/billboard/class_c5.txt'
att_l2_sa_path = r'/localnvme/data/billboard/attribute_l2_sa.yaml'
att_l2_sb_path = r'/localnvme/data/billboard/attribute_l2_sb.yaml'
att_l2_sc_path = r'/localnvme/data/billboard/attribute_l2_sc.yaml'
att_l2_sd_path = r'/localnvme/data/billboard/attribute_l2_sd.yaml'

train_test_path = os.path.join(data_c5l2_dir, 'train_test.txt')
val_test_path = os.path.join(data_c5l2_dir, 'val_test.txt')
train_80p_path = os.path.join(data_c5l2_dir, 'train_80p_ref.txt')
val_80p_path = os.path.join(data_c5l2_dir, 'val_80p_ref.txt')
all_path = os.path.join(data_c5l2_dir, 'all.txt')
all_test_path = os.path.join(data_c5l2_dir, 'all_test.txt')
```

```python
for risk in risk_list:
    extract_single_risk_keep_single(data_c5l2_dir_label, single_risk_dir, risk)
```

```python
for risk in risk_list:
    output_image_dir = os.path.join(single_risk_dir, risk, 'images')
    print(f'copy {output_image_dir}')
    shutil.rmtree(output_image_dir) if os.path.exists(output_image_dir) else None
    shutil.copytree(data_c5l2_dir_image, output_image_dir)
```

```python
for risk in risk_list:
    class_path = os.path.join(single_risk_dir, risk, 'class.txt')
    print(f'copy {class_path}')
    shutil.copy(class_c5_path, class_path)
    att_path = os.path.join(single_risk_dir, risk, 'attribute.yaml')
    print(f'copy {att_path}')
    if risk == 'a':
        shutil.copy(att_l2_sa_path, att_path)
    elif risk == 'b':
        shutil.copy(att_l2_sb_path, att_path)
    elif risk == 'c':
        shutil.copy(att_l2_sc_path, att_path)
    elif risk == 'd':
        shutil.copy(att_l2_sd_path, att_path)
    else:
        print('error')

    output_dir = os.path.join(single_risk_dir, risk)
    output_train_test_path = os.path.join(output_dir, 'train_test.txt')
    output_val_test_path = os.path.join(output_dir, 'val_test.txt')
    output_train_80p_path = os.path.join(output_dir, 'train_80p_ref.txt')
    output_val_80p_path = os.path.join(output_dir, 'val_80p_ref.txt')
    output_all_path = os.path.join(output_dir, 'all.txt')
    output_all_test_path = os.path.join(output_dir, 'all_test.txt')

    copy_split(train_test_path, output_train_test_path, abs_path=True)
    copy_split(val_test_path, output_val_test_path, abs_path=True)
    copy_split(train_80p_path, output_train_80p_path, abs_path=True)
    copy_split(val_80p_path, output_val_80p_path, abs_path=True)
    copy_split(all_path, output_all_path, abs_path=True)
    copy_split(all_test_path, output_all_test_path, abs_path=True)
```

```python
for risk in risk_list:
    ref_split(
        ref_path=r'/localnvme/data/billboard/all_data/mseg_c6/data7961_mseg_c6_1030/val_test_broken.txt',
        img_dir=os.path.join(single_risk_dir, risk, 'images'),
        label_dir=os.path.join(single_risk_dir, risk, 'labels'),
        add_suffix='_test_broken_syn_v1',
        full_path=True,
    )
```
