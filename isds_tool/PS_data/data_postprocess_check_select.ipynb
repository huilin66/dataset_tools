```python
import os
import shutil
import pandas as pd
import numpy as np
from tqdm.notebook import tqdm
```

```python
from logging import root


root_path = r'E:\data\202502_signboard\result'
class_file = r'E:\data\202502_signboard\data_annotation\docs\class_c6.txt'
att_file = r'E:\data\202502_signboard\data_annotation\docs\attribute.yaml'
yolo_merge_dir = os.path.join(root_path, 'merge_dir_0903')
yolo_merge_image_dir = os.path.join(yolo_merge_dir, 'images')
yolo_merge_label_dir = os.path.join(yolo_merge_dir, 'labels')
yolo_merge_label_re_dir = os.path.join(yolo_merge_dir, 'labels_re')
defect_list = ['deformation', 'broken', 'abandonment', 'corrosion']
task_name_list = [
        "task0808",
        "task0812",
    ]
task_num_list = [
    366,
    370
    ]
image_list_path = os.path.join(root_path, 'image_list.xlsx')
os.makedirs(yolo_merge_image_dir, exist_ok=True)
os.makedirs(yolo_merge_label_dir, exist_ok=True)
os.makedirs(yolo_merge_label_re_dir, exist_ok=True)
```

```python
def get_image_list(input_path):
    df = pd.read_excel(input_path, sheet_name='Sheet1')
    image_list = df['image_name'].astype(str).tolist()
    return list(set(image_list))
```

```python
img_list = get_image_list(image_list_path)
```

```python
for task_name, task_num in zip(task_name_list, task_num_list):
    input_dir = os.path.join(root_path, task_name, 'results', str(task_num), 'yolo_dataset')
    input_image_dir = os.path.join(input_dir, 'images')
    input_label_dir = os.path.join(input_dir, 'labels')
    
    count = 0
    image_list = os.listdir(input_image_dir)
    for image_name in tqdm(image_list):
        input_image_path = os.path.join(input_image_dir, image_name)
        input_label_path = os.path.join(input_label_dir, image_name.replace('.jpg', '.txt'))
        output_image_path = os.path.join(yolo_merge_image_dir, image_name)
        output_label_path = os.path.join(yolo_merge_label_dir, image_name.replace('.jpg', '.txt'))

        if image_name in img_list:
            shutil.copy(input_image_path, output_image_path) 
            shutil.copy(input_label_path, output_label_path)
            count +=1
    print(f'{task_name} {count}')
```

```python
label_list = os.listdir(yolo_merge_label_dir)
for label_name in tqdm(label_list):
    label_path = os.path.join(yolo_merge_label_dir, label_name)
    label_re_path = os.path.join(yolo_merge_label_re_dir, label_name)
    with open(label_path, 'r') as f_in, open(label_re_path, 'w+') as f_out:
        lines = f_in.readlines()
        for line in lines:
            num_list = line.strip().split(' ')
            num_list = num_list[:-1]
            line_seg = ' '.join(num_list) + '\n'
            f_out.write(line_seg)
```
