```python
import os
import pandas as pd
from tqdm import tqdm
from pathlib import Path
```

```python
obj_dir = r'E:\cp_dir\labels_crop\deformation\high'
obj_csv_path = obj_dir + r'_obj.csv'
```

```python
def obj2csv(input_dir, output_csv_path):
    df = pd.DataFrame(None, columns=['filename', 'cam', 'date', 'time', 'patrol', 'deformation'])
    file_list = os.listdir(input_dir)
    for file_name in tqdm(file_list):
        cam, timestamp, _ = Path(file_name).stem.split('_')
        date = timestamp[4:8]
        time = timestamp[8:12]
        df.loc[len(df)] = [file_name, cam, date, time, None, True]
    df.to_csv(output_csv_path, index=False)
```

```python
obj2csv(obj_dir, obj_csv_path)
```
