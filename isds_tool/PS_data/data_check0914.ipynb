```python
import os
import shutil
import pandas as pd
```

```python
csv_path = r'E:\data\202502_signboard\data_annotation\docs\check0914.csv'
csv_path_save = r'E:\data\202502_signboard\data_annotation\docs\check0914_split.csv'
```

```python
df = pd.read_csv(csv_path)
```

```python
def split_filename(filename):
    # 从右边分割一次 '_'，得到 "prefix_part" 和 "y.ext"
    prefix_part, y_ext = filename.rsplit('_', 1)
    
    # 再分割 "y.ext" 得到 "y" 和 "ext"
    y, ext = y_ext.split('.', 1)
    
    # 合并 "prefix_part" 和 "ext" 得到 "prefix"
    prefix = f"{prefix_part}.{ext}"
    
    return prefix, y
```

```python


# 应用函数并拆分成两列
df[['file_name', 'object_id']] = df['object_name'].apply(lambda x: pd.Series(split_filename(x)))
```

```python
df.to_csv(csv_path_save, index=False)
```
