```python
import cv2
import os
import glob

def images_to_video(image_folder, output_video_path, fps=10):
    # 1. 获取文件夹下所有图片的路径，并进行排序
    # 注意：确保文件名的排序规则与时间顺序一致（如 0001.jpg, 0002.jpg）
    image_paths = glob.glob(os.path.join(image_folder, "*.jpg")) # 如果是png格式请改为 *.png
    image_paths.sort() 
    
    if not image_paths:
        print("未找到图片，请检查路径！")
        return

    # 2. 读取第一张图片来获取视频的分辨率 (宽, 高)
    first_image = cv2.imread(image_paths[0])
    height, width, layers = first_image.shape
    size = (width, height)

    # 3. 初始化 VideoWriter
    # 'mp4v' 是 mp4 格式常用的编码器，也可以用 'XVID' 对应 .avi 格式
    fourcc = cv2.VideoWriter_fourcc(*'mp4v') 
    video = cv2.VideoWriter(output_video_path, fourcc, fps, size)

    print(f"开始合成视频，共 {len(image_paths)} 帧，帧率 {fps} FPS...")

    # 4. 逐帧写入视频
    for img_path in image_paths:
        img = cv2.imread(img_path)
        # 如果有些图片大小不一致，可以在这里加一步 resize，例如：
        # img = cv2.resize(img, size)
        video.write(img)

    # 5. 释放资源并保存
    video.release()
    print(f"视频已保存至: {output_video_path}")

# 使用示例
# 假设你的结果图片存在 'output_images' 文件夹下
images_to_video(
    r'\\158.132.186.40\isds\huilin\isds\google_drive_data\task1008\01-04-56_PM\cameras\camera5_vis', 
    r'\\158.132.186.40\isds\huilin\isds\google_drive_data\task1008\01-04-56_PM\cameras\camera5_vis_video.mp4', fps=10)
```
