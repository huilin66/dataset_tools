```python
subproject_id = "245"
download_directory = r"E:\data\202502_signboard\result\task0711"
data_dir = os.path.join(download_directory, subproject_id)
RESULT_ALL = "http://ec2-54-46-0-164.ap-east-1.compute.amazonaws.com/emes/api/v2/workflow-result/exportResults?subProjectId="
RESULT_JSON = "http://ec2-54-46-0-164.ap-east-1.compute.amazonaws.com/emes/api/v2/workflow-result/exportOutputResults?subProjectId="
```

```python
def download_zip_files(url_list, save_dir):
    os.makedirs(save_dir, exist_ok=True)

    session = requests.Session()
    session.headers.update({
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
    })

    for url in url_list:
        try:
            print(f"🔗 正在连接: {url}")
            response = session.get(url, stream=True)
            response.raise_for_status()

            total_size = int(response.headers.get("content-length", 0))
            chunk_size = 1024 * 1024  # 1MB

            print(f'total size: {total_size/chunk_size} MB')

            parsed_url = urlparse(url)
            query_params = parse_qs(parsed_url.query)
            sub_project_id = query_params.get('subProjectId', ['unknown'])[0]
            filename = f"{sub_project_id}.zip"
            save_path = os.path.join(save_dir, filename)

            with open(save_path, 'wb') as f, tqdm(
                total=total_size,
                unit='B',
                unit_scale=True,
                unit_divisor=1024,
                desc=f"📥 下载 {filename}",
                leave=True,
            ) as pbar:
                for chunk in response.iter_content(chunk_size=chunk_size):
                    if chunk:
                        f.write(chunk)
                        pbar.update(len(chunk))

            print(f"✅ 下载完成: {save_path}\n")

        except Exception as e:
            print(f"❌ 下载失败: {url}\n原因: {e}")
```

```python
zip_urls = [
    RESULT_ALL+subproject_id,
]

# 调用下载函数
download_zip_files(zip_urls, download_directory)
```

```python
def simple_unzip(zip_path, dst_dir):
    print(f'{zip_path} unzip...')
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(dst_dir)
    print(f'{zip_path} done\n')
```

```python
zip_path = os.path.join(download_directory, f'{subproject_id}.zip')
dst_dir = os.path.join(download_directory, subproject_id)
simple_unzip(zip_path, dst_dir=dst_dir)
```
