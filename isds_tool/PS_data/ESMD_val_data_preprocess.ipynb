```python
import os
import cv2
import numpy as np
from tqdm import tqdm
from pathlib import Path
from es_tools import DataUploader
from zip_tools import zip_folder_to_path
```

```python
root_dir = r'\\158.132.186.40\isds\huilin\isds\other_data\upload1014\Val_set\Val_set'
```

```python
sub_list = [
    'cam1',
    'cam2',
    'cam3',
    'cam4',
    'cam5',
    'cam6',
]
cam_list = [
    'DA5148683',
    'DA5324655',
    'DA4930148',
    'DA5324645',
    'DA5148680',
    'DA6102933',
]
```

```python
def create_empty_img(width, height, save_path):
    img = np.zeros((height, width, 3), dtype=np.uint8)
    cv2.imwrite(save_path, img)
```

```python
width, height = None, None
```

```python
timestamp_list = []
for sub_name in sub_list:
    sub_dir = os.path.join(root_dir, sub_name)
    for file_name in tqdm(os.listdir(sub_dir), desc=f'collect {sub_name} timestamp'):
        if width is None:
            height, width = cv2.imread(os.path.join(sub_dir, file_name)).shape[:2]
        timestamp = Path(file_name).stem.split('_')[1]
        timestamp_list.append(timestamp)
timestamp_list = list(set(timestamp_list))
```

```python
print(width, height)
```

```python
for timestamp in tqdm(timestamp_list, desc='process timestamp'):
    for cam_name, sub_name in zip(cam_list, sub_list):
        file_name = f'{cam_name}_{timestamp}.jpg'
        file_path = os.path.join(root_dir, sub_name, file_name)
        if not os.path.exists(file_path):
            print(f'creating {file_path}')
            create_empty_img(width, height, file_path)
    
```

```python
for sub_name in sub_list:
    zip_folder_to_path(
        source_folder=os.path.join(root_dir, sub_name),
        destination_zip=os.path.join(root_dir, sub_name+'.zip')
    )
```
