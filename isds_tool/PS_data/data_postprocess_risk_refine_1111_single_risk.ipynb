```python
for risk in risk_list:
    output_image_dir = os.path.join(single_risk_dir, risk, 'images')
    print(f'copy {output_image_dir}')
    shutil.copytree(datav15_c5l2_dir_image, output_image_dir)
```

```python
for risk in risk_list:
    class_path = os.path.join(single_risk_dir, risk, 'class.txt')
    print(f'copy {class_path}')
    shutil.copy(class_c5_path, class_path)
    att_path = os.path.join(single_risk_dir, risk, 'attribute.yaml')
    print(f'copy {att_path}')
    if risk == 'a':
        shutil.copy(att_l2_sa_path, att_path)
    elif risk == 'b':
        shutil.copy(att_l2_sb_path, att_path)
    elif risk == 'c':
        shutil.copy(att_l2_sc_path, att_path)
    elif risk == 'd':
        shutil.copy(att_l2_sd_path, att_path)
    else:
        print('error')

    output_dir = os.path.join(single_risk_dir, risk)
    output_train_test_path = os.path.join(output_dir, 'train_test.txt')
    output_val_test_path = os.path.join(output_dir, 'val_test.txt')
    output_train_80p_path = os.path.join(output_dir, 'train_80p_ref.txt')
    output_val_80p_path = os.path.join(output_dir, 'val_80p_ref.txt')
    output_all_path = os.path.join(output_dir, 'all.txt')
    output_all_test_path = os.path.join(output_dir, 'all_test.txt')

    copy_split(train_test_path, output_train_test_path, abs_path=True)
    copy_split(val_test_path, output_val_test_path, abs_path=True)
    copy_split(train_80p_path, output_train_80p_path, abs_path=True)
    copy_split(val_80p_path, output_val_80p_path, abs_path=True)
    copy_split(all_path, output_all_path, abs_path=True)
    copy_split(all_test_path, output_all_test_path, abs_path=True)
```
